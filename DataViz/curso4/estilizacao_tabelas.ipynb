```python
import pandas as pd
```

```python
df = pd.read_csv("./data/relatorio_vendas.csv")
df.sample(5)
```

```python
df.data_pedido = pd.to_datetime(df.data_pedido)
```

```python
df.info()
```

## Time de Vendas

> **Top 10 clientes por total em Vendas**

```python
df_cliente_vendas = df.groupby("nome_cliente")["vendas"].sum().nlargest(10).copy()
df_cliente_vendas = df_cliente_vendas.reset_index()
df_cliente_vendas
```

```python
df_cliente_vendas.columns = ["Clientes", "Vendas"]

# Col Auxiliar
df_cliente_vendas["Ranque"] = df_cliente_vendas.index + 1
df_cliente_vendas = df_cliente_vendas.set_index("Ranque")
display(df_cliente_vendas)
```

```python
table_top10 = df_cliente_vendas.style
table_top10
```

```python
print(type(table_top10))
```

```python
# Formatação R$

table_top10.format(
    {"Vendas": "R$ {:,.2f}"} 
)
table_top10
```

## 02 - Destacando Elementos

O total ganho em vendas e o total ganho em lucros a partir dos tipos de produtos vendidos.

```python
df_venda_lucro = df.groupby("tipo_produto")[["vendas", "lucro"]].sum()
df_venda_lucro.index.name = "tipo produto"
df_venda_lucro
```

```python
# Criando objeto style
table_venda_lucro = df_venda_lucro.style
```

```python
table_venda_lucro.format("R${:,.2f}")
table_venda_lucro.highlight_max(color="green").highlight_min(color="#F16165")
```

### Background em Gradiente 

```python
table_venda_lucro.background_gradient(cmap="Greens")
```
